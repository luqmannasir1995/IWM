```python
import numpy as np
import pandas as pd
import pandas_datareader.data as web
import statsmodels.api as sm
import statsmodels.formula.api as smf
from linearmodels import PanelOLS, FamaMacBeth
from tqdm.notebook import tqdm
import random
random.seed(809)

import datetime
from datetime import timedelta
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

# Load the Data
## Factors 

```python
FF3 = pd.read_csv(r'F-F_Research_Data_Factors_daily.CSV', skiprows = 3)
FF3.rename(columns = {'Unnamed: 0':'Date'}, inplace = True)
FF3['Date'] = FF3['Date'].astype(str)
FF3.rename(columns={'Mkt-RF':'Mkt_RF'}, inplace = True)
FF3['Mkt_RF'] = FF3['Mkt_RF']/100
FF3['SMB'] = FF3['SMB']/100
FF3['HML'] = FF3['HML']/100
FF3['RF'] = FF3['RF']/100
FF3 = FF3[~FF3['Mkt_RF'].isna()]
FF3
```

## TSLA, AAPL, MSFT and GOOGL Stocks Return

```python
Stocks = pd.read_csv('Stocks.csv', low_memory=False)
Stocks['RET'] = pd.to_numeric(Stocks['RET'], errors = 'coerce')
Stocks['Date'] = Stocks['date'].astype(str)

Stocks.drop(columns = ['date', 'PERMNO'], inplace = True)
Stocks.dropna(inplace = True)
Stocks.sort_values(['TICKER', 'Date'], inplace = True)
Stocks.drop_duplicates(subset = ['TICKER', 'Date'], inplace= True, keep =False)
```

```python
Stocks_FF3 = pd.merge(Stocks, FF3, on ='Date', how = 'left')
Stocks_FF3['Date']=pd.to_datetime(Stocks_FF3['Date'], format = '%Y%m%d')
```

```python
Stocks_FF3['RET_RF'] = Stocks_FF3['RET'] - Stocks_FF3['RF']
Stocks_FF3
```

### We will study the returns after the 2019Q4 earnings release

```python
event_date = {'TSLA': datetime.date(2019,10,23), 
              'MSFT': datetime.date(2019,10,23), 
              'AAPL': datetime.date(2019,10,30), 
              'GOOGL': datetime.date(2019,10,28)
             }
```

# Step 1: Pick a model of expected returns and calculate “abnormal returns” (AR) for each security, for each day around the event:

## Model: 
- Fama-French 3 Factors


## Train the model: 
- Using the last 3 months of data prior to the event, 
- Up to 10 days before the event

```python
start_training = 100
end_training = 10
```

```python
results ={}
coefs={}

for asset in tqdm(Stocks_FF3['TICKER'].unique()):  
    
    #Select the right asset
    df = Stocks_FF3[Stocks_FF3.TICKER ==asset]
    #Select the event
    event = event_date[asset]
    
    #Select the right period for the returns
    df = df[(df['Date']> pd.to_datetime(event-timedelta(days=start_training))) &
            (df['Date']<=pd.to_datetime(event-timedelta(days=end_training)))]
            
    #Run the Regression
    reg = smf.ols(formula='RET_RF ~ 1 + Mkt_RF + SMB + HML', data=df)
    res = reg.fit()

    #Record the resukt and the coefficients
    results[asset] = res.summary()
    coefs[asset] = res.params
```

## Applying the model

```python
CARs = {}
for asset in tqdm(Stocks_FF3['TICKER'].unique()):    
    
    #Select the right asset
    df = Stocks_FF3[Stocks_FF3.TICKER ==asset]
    
    #Select the event
    event = event_date[asset]
    
    #Select the right period for the returns: 10 days before the event an, up to 30 days after:
    df = df[(df.Date>pd.to_datetime(event-timedelta(days=10))) & 
            (df.Date<=pd.to_datetime(event+timedelta(days=30)))]

    df['E_R'] = coefs[asset]['Intercept'] + \
                        coefs[asset]['Mkt_RF']* df['Mkt_RF'] + \
                        coefs[asset]['SMB']* df['SMB'] + \
                        coefs[asset]['HML']* df['HML'] 
    
    #Create the abnormal return
    df['AR'] = df['RET']-df['E_R']
    
    #Create the cumulative abnormal return
    df['CAR'] = df['AR'].cumsum()
    
    #Create window dates
    df['Event_time'] = pd.to_datetime(df['Date'])-pd.to_datetime(event_date[asset])
    
    #Merge back
    df = df[['TICKER', 'Date', 'Event_time', 'AR', 'CAR']]
    CARs[asset] = df
    
```

```python
CARs['TSLA'][['TICKER', 'Date', 'Event_time', 'AR', 'CAR']]
```

# Plot

```python
Comp ='TSLA'

fig, ax = plt.subplots(figsize=(10,8))
ax.plot(CARs[Comp]['Date'], CARs[Comp]['CAR'], label = 'CAR', marker='o')
ax.xaxis.set_tick_params(rotation=45)
ax.set_title('Cumulative Abnormal Returns for '+Comp)
ax.set_xlabel('Date')
ax.set_ylabel('CAR')
plt.axvline(event_date[Comp], color='orange', label = 'Earnings Release Date')
ax.legend(loc = 'lower right')
plt.show()
```
